# Module X: Vision Agent
This controlled demonstration supplies synthetic structured outputs from existing modules. The agent does not analyze a raw image, diagnose, or estimate disease probability.

```python
from spectraderm.agents.vision_agent import VisionAgent, VisionAgentInput

# Synthetic, controlled outputs only; no patient image or API key is used.
image_quality = {'status': 'GOOD', 'passed': True}
localization = {'method': 'synthetic_roi_localization', 'candidates': ['synthetic-roi']}
reconstruction = {'estimated': True, 'output_bands': 31}
spectral_features = {'features': {'spectral.synthetic_ratio': 1.05}}
monitoring_change = {'category': 'Monitor', 'change_score': 52.0}

agent_input = VisionAgentInput(
    image_quality_result=image_quality,
    localization_result=localization,
    reconstruction_result=reconstruction,
    spectral_features=spectral_features,
    monitoring_change_result=monitoring_change,
)
```

```python
result = VisionAgent().interpret(agent_input)

print('IMAGE-QUALITY RESULT:', image_quality)
print('LOCALIZATION RESULT:', localization)
print('RECONSTRUCTION RESULT:', reconstruction)
print('SPECTRAL FEATURES:', spectral_features)
print('MONITORING/CHANGE RESULT:', monitoring_change)
print('\nVISION AGENT INPUT:', agent_input)
print('\nVISION AGENT OUTPUT:', result)
print('\nStructured finding for an optional later Module W handoff:', result.structured_finding)
```
