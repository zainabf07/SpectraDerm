# Module AA: Safety Agent
All examples are synthetic. AA is a deterministic presentation-safety gate, not a diagnostic system or medical-risk calculator.

```python
from spectraderm.agents.safety_agent import SafetyAgent, SafetyAgentInput
from spectraderm.rag.rag_pipeline import RetrievedEvidence, SourceMetadata

def make_evidence(text):
    return {'retrieved_evidence': (RetrievedEvidence('synthetic:chunk-0001', text, 1.0, SourceMetadata('synthetic', 'Synthetic source', 'Synthetic organization', 'https://example.test', 'synthetic context')),), 'grounding_status': 'evidence_found', 'explanation': 'Model-derived finding is separate from retrieved evidence.'}

agent = SafetyAgent()
```

```python
scenarios = {
  'SAFE': SafetyAgentInput(finding_summary='Visible pigmentation-related feature change detected.', evidence_result=make_evidence('Relevant retrieved knowledge provides appearance context.')),
  'UNSUPPORTED DIAGNOSIS': SafetyAgentInput(finding_summary='Model output', evidence_result=make_evidence('Relevant context.'), proposed_action='This proves the person has acne.'),
  'INCREASING MODEL CHANGE': SafetyAgentInput(finding_summary='Model-derived change score increased.', monitoring_result={'trend': 'increasing_change'}, evidence_result=make_evidence('Retrieved context without a referral claim.')),
  'INSUFFICIENT EVIDENCE': SafetyAgentInput(finding_summary='Model output', evidence_result={'retrieved_evidence': (), 'grounding_status': 'insufficient_evidence', 'explanation': ''}),
}
for name, item in scenarios.items():
    result = agent.evaluate(item)
    print('\nSCENARIO:', name)
    print('INPUT:', item)
    print('STATUS:', result.safety_status, '| professional assessment:', result.professional_assessment_recommended)
    print('BLOCKED:', result.blocked_claims)
    print('REASONS:', result.reasons)
    print('SAFETY:', result.safety_message)
```
