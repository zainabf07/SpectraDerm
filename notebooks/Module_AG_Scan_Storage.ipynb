# Module AG — Scan Storage

This demonstration uses a synthetic pseudonymous user ID, synthetic scan references, and a temporary private storage root. No real images, user information, or coordinates are used.

```python
from pathlib import Path
from tempfile import TemporaryDirectory
from spectraderm.storage.scan_storage import ScanStorage

class SyntheticAFCompatibleUsers:
    def get_user(self, user_id):
        return {'user_id': user_id} if user_id == 'usr_synthetic_demo' else None

temporary_root = TemporaryDirectory()
storage = ScanStorage(Path(temporary_root.name), SyntheticAFCompatibleUsers())
user_id = 'usr_synthetic_demo'
```

## Create synthetic longitudinal scans

References are relative identifiers only; AG stores no image bytes in the metadata record.

```python
scan_one = storage.create_scan(user_id, 'synthetic/image-001.ref', feature_reference='synthetic/features-001.ref')
scan_two = storage.create_scan(user_id, 'synthetic/image-002.ref', analysis_reference='synthetic/analysis-002.ref')
scan_one, scan_two
```

## Chronological history and ownership-checked retrieval

```python
storage.list_user_scans(user_id)
storage.get_scan(user_id, scan_one.scan_id)
```

## Delete one scan and verify deletion

```python
storage.delete_scan(user_id, scan_one.scan_id)
storage.get_scan(user_id, scan_one.scan_id), storage.list_user_scans(user_id)
```
