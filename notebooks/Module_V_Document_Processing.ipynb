# Module V: Document Processing
This notebook demonstrates deterministic local document processing and validation. It does not provide medical interpretation, diagnosis, or treatment guidance.

```python
from spectraderm.rag.document_processing import (
    clean_documents, load_source_documents, process_knowledge_base, validate_chunks,
)

documents = load_source_documents()
for document in documents:
    print(document.source.source_id, '-', document.path.name)
```

```python
cleaned_documents = clean_documents(documents)
chunks = process_knowledge_base()
validation = validate_chunks(chunks, tuple(item.source for item in cleaned_documents))

print('Sources:', len(cleaned_documents))
print('Chunks:', len(chunks))
print('Validation:', validation)

for chunk in chunks[:3]:
    print('\n', chunk.chunk_id, '|', chunk.section)
    print(chunk.text[:240])
```
