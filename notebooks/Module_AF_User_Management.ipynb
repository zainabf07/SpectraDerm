# Module AF — User Management

This notebook uses only synthetic pseudonymous IDs and synthetic scan IDs. AF stores no image contents, names, contact details, or precise coordinates.

```python
from spectraderm.users.user_management import UserManagement

users = UserManagement()
user = users.create_user(general_consent=True, location_consent=False)
user
```

## Independent consent controls

Changing general consent does not imply location consent, and vice versa.

```python
users.update_location_consent(user.user_id, True)
users.get_user(user.user_id)
```

## Scan-ID association only

These are synthetic identifiers; AF does not accept or store scan/image contents.

```python
users.add_scan(user.user_id, 'synthetic-scan-001')
users.add_scan(user.user_id, 'synthetic-scan-002')
users.get_user_scans(user.user_id)
```

## Deletion

Deleting the AF user removes its AF record and scan associations.

```python
users.delete_user(user.user_id), users.get_user(user.user_id)
```
