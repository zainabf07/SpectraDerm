# Module R: Personal Feature Baseline
This is a software/integration demonstration using synthetic Module N-style feature mappings. It is not clinical validation.

```python
from spectraderm.monitoring.baseline import build_personal_baseline

history = [
    {"rgb.color_mean_r": 10.0, "spectral.signature_540": 0.30},
    {"rgb.color_mean_r": 12.0, "spectral.signature_540": 0.35},
    {"rgb.color_mean_r": 11.0, "spectral.signature_540": 0.32},
]
baseline = build_personal_baseline(history)
```

```python
print('Observation count:', baseline.observation_count)
print('Unavailable features:', baseline.unavailable_features)
print('Baseline feature values:')
for name, value in baseline.baseline_features.items():
    print(f'  {name}: {value} (contributors: {baseline.feature_contributions[name]})')
```

```python
# Non-finite values do not become zeros; this feature remains unavailable.
handled = build_personal_baseline([
    {"rgb.valid": 1.0, "spectral.unavailable": float('nan')},
    {"rgb.valid": 3.0, "spectral.unavailable": float('inf')},
])
print(handled)
```
