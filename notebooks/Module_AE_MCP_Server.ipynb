# Module AE — MCP Server

All examples use injected mock services. **Synthetic provider data — Google integration is implemented later in Module AI.**

```python
from spectraderm.mcp.server import MCPCapabilities, create_mcp_server

class SyntheticProvider:
    def find_dermatologists(self, location, radius_km, specialty='dermatology'):
        return ({'name': 'Synthetic Dermatology Clinic', 'specialty': specialty, 'source': 'synthetic'},)

capabilities = MCPCapabilities(
    calculate_warning_score=lambda payload: {'engineering_warning': 'synthetic', 'input': payload},
    retrieve_evidence=lambda payload: {'evidence': 'synthetic supplied result', 'input': payload},
    dermatologist_provider=SyntheticProvider(),
)
server = create_mcp_server(capabilities)
server
```

## Initialization, registered tools, and example schemas

```python
server.tool_names
server.tool_schemas['find_dermatologists']
```

## Representative injected tools

These outputs are structured MCP envelopes; no medical reasoning is performed by AE.

```python
server.call_tool('calculate_warning_score', input={'existing_module_q_output': 'synthetic'})
server.call_tool('retrieve_evidence', input={'query': 'synthetic supplied query'})
server.call_tool('find_dermatologists', location={'city': 'Synthetic City'}, radius_km=10)
server.call_tool('get_otc_product_categories')
```

Product output is category-level only: no brands, partner products, prices, or affiliate links. The MCP server delegates decisions about appropriate categories to existing specialized logic.
