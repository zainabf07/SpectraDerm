# Module Y: Monitoring Agent
This notebook uses synthetic engineering records only. An increasing change trend means an increasing model-derived change score, not disease progression.

```python
from spectraderm.agents.monitoring_agent import HistoricalScanRecord, MonitoringAgent, MonitoringAgentInput

# Synthetic sequence: baseline, small change, similar change, higher change.
history = (
    HistoricalScanRecord('scan-1', '2026-01-01T09:00:00', 10.0, 'Stable', 'synthetic baseline'),
    HistoricalScanRecord('scan-2', '2026-02-01T09:00:00', 12.0, 'Monitor', 'synthetic small change'),
    HistoricalScanRecord('scan-3', '2026-03-01T09:00:00', 13.0, 'Monitor', 'synthetic similar change'),
    HistoricalScanRecord('scan-4', '2026-04-01T09:00:00', 20.0, 'Higher Change', 'synthetic higher change'),
)
current_q_output = {'change_score': 20.0, 'category': 'Higher Change'}
agent_input = MonitoringAgentInput(
    baseline_reference={'source': 'synthetic Module R output'},
    current_comparison={'source': 'synthetic Module S output'},
    change_anomaly_result=current_q_output,
    historical_scan_records=history,
    current_scan_id='scan-4',
 )
```

```python
result = MonitoringAgent().interpret(agent_input)
print('HISTORICAL RECORDS:', result.historical_records)
print('CURRENT SCORE:', result.current_change_score)
print('PREVIOUS SCORE:', result.previous_change_score)
print('TREND:', result.trend)
print('FINAL MONITORING RESULT:', result)
print('\nBoundary: increasing_change is an increasing model-derived score, not disease progression.')
```
