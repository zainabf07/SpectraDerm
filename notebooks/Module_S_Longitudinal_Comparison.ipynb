# Module S: Longitudinal Comparison
Software/integration demonstration with controlled compatible values only; this is not clinical validation.

```python
import numpy as np
from spectraderm.features.combined_features import CombinedFeatureResult, FeatureMetadata
from spectraderm.monitoring.baseline import build_personal_baseline
from spectraderm.monitoring.longitudinal import compare_longitudinal

baseline = build_personal_baseline([
    {"rgb.color_mean_r": 10.0, "spectral.signature_540": 0.30},
    {"rgb.color_mean_r": 12.0, "spectral.signature_540": 0.34},
])
current_features = {"rgb.color_mean_r": 13.0, "spectral.signature_540": 0.32}
current = CombinedFeatureResult(
    current_features=current_features, temporal_features={}, combined_features=dict(current_features),
    ml_vector=np.asarray(list(current_features.values()), dtype=float), feature_names=list(current_features),
    metadata=FeatureMetadata("demo-current", None, None, True, "follow_up", None, None, True, 1), warnings=[]
)
comparison = compare_longitudinal(baseline, current)
```

```python
print('Summary:', comparison.summary)
print('Metadata:', comparison.metadata)
for name in comparison.feature_status:
    print(name, comparison.feature_status[name], 'delta=', comparison.delta[name], 'relative=', comparison.relative_change[name])
```
