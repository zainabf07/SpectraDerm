# Module L - RGB Feature Extraction

A local, deterministic demonstration of conventional RGB feature extraction for a selected region of interest (ROI).

## Objective

Extract interpretable color, texture, shape, local ROI contrast, and RGB appearance-proxy features. This notebook deliberately uses a small synthetic image and mask so it does not require the Hyper-Skin archive.

```python
from pathlib import Path
import sys

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

root = Path.cwd().resolve()
if not (root / 'src').exists():
    root = root.parent
sys.path.insert(0, str(root / 'src'))

from spectraderm.features.rgb_features import extract_rgb_features
```

## Load RGB image and use an ROI mask

Replace this synthetic example with a locally available RGB image and a mask supplied by an upstream workflow when appropriate. The mask below is a **synthetic demonstration ROI**, not a disease or lesion label.

```python
rgb = np.full((96, 96, 3), [150, 105, 80], dtype=np.uint8)
rgb[32:64, 32:64] = [180, 120, 90]
mask = np.zeros((96, 96), dtype=bool)
mask[32:64, 32:64] = True

fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(rgb); axes[0].set_title('Synthetic RGB image'); axes[0].axis('off')
axes[1].imshow(mask, cmap='gray'); axes[1].set_title('Synthetic demonstration ROI'); axes[1].axis('off')
plt.show()
```

## Run RGB feature extraction

```python
result = extract_rgb_features(rgb, mask)
result
```

## Display feature groups and feature table

```python
for group, names in result.feature_groups.items():
    print(f'{group}: {names}')

feature_table = pd.DataFrame(
    {'feature': result.features.keys(), 'value': result.features.values()}
).set_index('feature')
feature_table
```

## Brief interpretation

Color and texture summarize selected pixel appearance. Shape describes only the supplied mask. Local contrast compares the ROI to nearby image context. Pigmentation-related values are RGB appearance proxies, not direct biological measurements.

## Safety and limitations

RGB feature extraction does not establish disease presence or absence. It does not diagnose acne, inflammation, pigment disorders, lesions, cancer, or other medical conditions. Results depend on illumination, camera processing, and the supplied ROI mask.
