# Module Q: Subtle Skin-Pattern Change / Anomaly Analysis
A lightweight, synthetic demonstration. The Change/Anomaly Score measures feature-pattern difference from selected reference history; it is not a disease probability.

```python
from spectraderm.monitoring.anomaly import analyze_change

def observation(rgb, spectral, fixed=1.0):
    return {"rgb.texture": rgb, "spectral.signature_540": spectral, "rgb.fixed": fixed}

history = [observation(8, 80), observation(10, 100), observation(12, 120), observation(14, 140)]
history
```

```python
def display_result(label, result):
    print(label)
    print('RGB change:', result.rgb_change)
    print('Spectral change:', result.spectral_change)
    print('Change/Anomaly Score:', result.change_score)
    print('Category:', result.category)
    print('Feature coverage:', f'{result.features_compared}/{result.features_available}')
    print('Warnings:', result.warnings)

stable = analyze_change(observation(11, 110), history)
display_result('Stable case', stable)
```

```python
changed = analyze_change(observation(30, 300), history)
display_result('Changed case', changed)

insufficient = analyze_change(observation(11, 110), history[:3])
display_result('Insufficient-history case', insufficient)
```

```python
# rgb.fixed has zero reference IQR; spectral.bad is non-finite in current.
current = {"rgb.fixed": 99.0, "spectral.signature_540": 110.0, "spectral.bad": float('nan')}
references = [{"rgb.fixed": 1.0, "spectral.signature_540": x, "spectral.bad": 2.0} for x in (80, 100, 120, 140)]
handled = analyze_change(current, references)
display_result('Non-finite and zero-variance handling', handled)
```
