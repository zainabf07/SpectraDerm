# Module W: RAG Retrieval and Evidence-Grounded LLM
This demonstration keeps model-derived findings, retrieved medical evidence, and generated language separate. It is not diagnosis, medical advice, or disease prediction.

```python
from spectraderm.rag.llm import DeterministicMockLLM
from spectraderm.rag.rag_pipeline import RAGPipeline
from spectraderm.rag.retriever import VectorRetriever

# Requires a locally built Module U Step 2 index; no LLM API key is used.
retriever = VectorRetriever.load()
provider = DeterministicMockLLM('Mock explanation based only on the supplied evidence.')
pipeline = RAGPipeline(retriever, provider)
```

```python
model_finding = 'Existing model-derived appearance-change description.'
result = pipeline.run(model_finding, top_k=3)

print('MODEL-DERIVED FINDING:', result.model_finding)
print('RETRIEVAL QUERY:', result.query)
print('\nRETRIEVED MEDICAL EVIDENCE:')
for evidence in result.retrieved_evidence:
    print(evidence.chunk_id, evidence.source.source_id, evidence.source.url)
    print(evidence.text[:220])
print('\nGENERATED EXPLANATION:', result.explanation)
print('SAFETY:', result.safety)
```
