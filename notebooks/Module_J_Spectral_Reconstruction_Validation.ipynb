```python
from pathlib import Path
from PIL import Image
import h5py
import numpy as np

test_root = Path("/content/hyperskin_test3/Hyper-Skin(RGB, VIS)/test")

rgb_path = test_root / "RGB" / "p035_neutral_front.jpg"
vis_path = test_root / "VIS" / "p035_neutral_front.mat"

# Load RGB
rgb = np.asarray(Image.open(rgb_path)).astype(np.float32) / 255.0

# Load VIS ground truth
with h5py.File(vis_path, "r") as f:
    vis = np.asarray(f["cube"], dtype=np.float32)

print("RGB shape:", rgb.shape)
print("RGB dtype:", rgb.dtype)
print("RGB range:", rgb.min(), "to", rgb.max())

print("VIS raw shape:", vis.shape)
print("VIS dtype:", vis.dtype)
print("VIS range:", vis.min(), "to", vis.max())
```

```python
import torch
import numpy as np

# Use the top-left 128x128 region.
rgb_crop = rgb[:128, :128, :]

# Corresponding ground-truth VIS crop.
vis_gt = vis[:, :128, :128]

# HWC -> BCHW
rgb_tensor = torch.from_numpy(
    np.transpose(rgb_crop, (2, 0, 1))[None, ...]
).float().to(device)

print("RGB tensor:", rgb_tensor.shape)
print("Ground truth VIS:", vis_gt.shape)

# MST++ prediction
with torch.inference_mode():
    prediction = model(rgb_tensor)

prediction = prediction.detach().cpu().numpy()[0]

print("Predicted VIS:", prediction.shape)
print("Prediction finite:", np.isfinite(prediction).all())
```

```python
# Calculate Mean Absolute Error (MAE)

absolute_error = np.abs(prediction - vis_gt)

mae = np.mean(absolute_error)

print("MAE:", float(mae))
```

```python
band_mae = np.mean(
    np.abs(prediction - vis_gt),
    axis=(1, 2),
)

print("Per-band MAE:")
for i, value in enumerate(band_mae):
    wavelength = 400 + i * 10
    print(f"{wavelength} nm: {value:.6f}")
```

```python
rmse = np.sqrt(
    np.mean((prediction - vis_gt) ** 2)
)

print("RMSE:", float(rmse))
```

```python
band_rmse = np.sqrt(
    np.mean((prediction - vis_gt) ** 2, axis=(1, 2))
)

print("Per-band RMSE:")
for i, value in enumerate(band_rmse):
    wavelength = 400 + i * 10
    print(f"{wavelength} nm: {value:.6f}")
```

```python
# Spectral Angle Mapper (SAM)
numerator = np.sum(prediction * vis_gt, axis=0)

pred_norm = np.sqrt(np.sum(prediction ** 2, axis=0))
gt_norm = np.sqrt(np.sum(vis_gt ** 2, axis=0))

cosine = numerator / (pred_norm * gt_norm + 1e-8)
cosine = np.clip(cosine, -1.0, 1.0)

sam_map = np.arccos(cosine)

mean_sam = np.mean(sam_map)

print("Mean SAM (radians):", float(mean_sam))
print("Mean SAM (degrees):", float(np.degrees(mean_sam)))
```
