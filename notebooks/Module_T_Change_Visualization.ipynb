# Module T: Change Visualization
A software/integration demonstration with controlled feature values. The view model is not clinical validation or a diagnosis.

```python
import numpy as np
from spectraderm.features.combined_features import CombinedFeatureResult, FeatureMetadata
from spectraderm.monitoring.anomaly import analyze_change
from spectraderm.monitoring.baseline import build_personal_baseline
from spectraderm.monitoring.longitudinal import compare_longitudinal
from spectraderm.monitoring.visualization import build_change_visualization

history = [
    {"rgb.color_mean_r": 10.0, "spectral.signature_540": 0.30},
    {"rgb.color_mean_r": 12.0, "spectral.signature_540": 0.34},
    {"rgb.color_mean_r": 14.0, "spectral.signature_540": 0.38},
    {"rgb.color_mean_r": 16.0, "spectral.signature_540": 0.42},
]
current_features = {"rgb.color_mean_r": 15.0, "spectral.signature_540": 0.36}
current = CombinedFeatureResult(current_features, {}, dict(current_features), np.asarray(list(current_features.values())), list(current_features), FeatureMetadata("demo", None, None, True, "follow_up", None, None, True, 1), [])
baseline = build_personal_baseline(history)
comparison = compare_longitudinal(baseline, current)
visualization = build_change_visualization(baseline, comparison, analyze_change(current, history))
```

```python
print(visualization.labels)
print(visualization.score_panel)
for row in visualization.feature_rows:
    print(row.feature_name, row.direction, row.delta, row.status)
```
