# Module AI — Dermatologist Database/API

All examples use synthetic/mock provider data. No external request is made. No real API key is included.

```python
from spectraderm.referral.dermatologist_api import DermatologistSearchService, GooglePlacesProvider

class SyntheticProvider:
    def __init__(self, results=(), fail=False): self.results, self.fail = results, fail
    def search_dermatologists(self, location, radius_meters, limit):
        if self.fail: raise RuntimeError('synthetic provider failure')
        return self.results[:limit]

provider = SyntheticProvider(({'place_id': 'synthetic-1', 'name': 'Synthetic Dermatology Clinic', 'formatted_address': '1 Demo Road', 'types': ['dermatology']},))
service = DermatologistSearchService(provider, source='synthetic')
```

## Manual-location search and normalized records

The search location is a transient input; it is not persisted by the service.

```python
service.search_dermatologists({'city': 'Synthetic City'}, radius_meters=5000, limit=5)
```

## Empty results and provider failure

```python
DermatologistSearchService(SyntheticProvider(), source='synthetic').search_dermatologists({'city': 'Empty City'})
DermatologistSearchService(SyntheticProvider(fail=True), source='synthetic').search_dermatologists({'city': 'Failure City'})
```
