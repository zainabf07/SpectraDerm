# Module U Step 2: Embeddings and Vector Retrieval
This notebook retrieves source-attributed evidence candidates only. It does not generate medical answers, diagnoses, or clinical conclusions.

```python
from spectraderm.rag.knowledge_base import build_chunks, load_sources
from spectraderm.rag.retriever import build_index, load_or_build_retriever

print('Sources:', len(load_sources()))
print('Chunks:', len(build_chunks()))
metadata = build_index()  # Downloads/uses FastEmbed model cache locally if needed.
print(metadata)
```

```python
retriever = load_or_build_retriever()
queries = [
    'pigmentation and melanin',
    'visible redness and vascular appearance',
    'professional assessment for warning signs',
]
for query in queries:
    print('\nQuery:', query)
    for result in retriever.retrieve(query, top_k=3):
        print(result.score, result.title, result.url, result.chunk_id)
```
