# Module AC — Referral Agent

**Synthetic provider data for demonstration only.** This notebook makes no external API, Google, device-location, model, or image call.

```python
from spectraderm.agents.referral_agent import (
    DermatologistOption, ReferralAgent, ReferralAgentInput,
)

class SyntheticProvider:
    """Test/demo substitute for a future external provider."""
    def __init__(self):
        self.calls = []

    def find_dermatologists(self, location, radius_km, specialty='dermatology'):
        self.calls.append((dict(location), radius_km, specialty))
        return [
            DermatologistOption('Synthetic Far Skin Clinic', '30 Demo Road', 'Dermatology', 7.5, source='synthetic'),
            DermatologistOption('Synthetic Near Skin Clinic', '10 Demo Road', 'Dermatology', 2.0, 'Mon-Fri 09:00-17:00', source='synthetic'),
            DermatologistOption('Synthetic General Clinic', '5 Demo Road', 'General clinic', 1.0, source='synthetic'),
        ]
```

## Case 1 — professional assessment recommended

Synthetic provider data is ranked deterministically: dermatology category match, shorter returned distance, then available opening-hours information.

```python
provider = SyntheticProvider()
agent = ReferralAgent(provider)
case_1 = agent.refer(ReferralAgentInput(
    safety_result={'professional_assessment_recommended': True},
    location={'city': 'Synthetic City'},
))
case_1, provider.calls
```

## Case 2 — professional assessment not recommended

The provider is not called and no referral options are returned.

```python
no_referral_provider = SyntheticProvider()
case_2 = ReferralAgent(no_referral_provider).refer(ReferralAgentInput(
    safety_result={'professional_assessment_recommended': False},
    location={'city': 'Synthetic City'},
))
case_2.status, case_2.options, no_referral_provider.calls
```

## Case 3 — referral required, no location supplied

The agent asks for location permission or a manual location; it does not access a device itself.

```python
missing_location_provider = SyntheticProvider()
case_3 = ReferralAgent(missing_location_provider).refer(ReferralAgentInput(
    safety_result={'professional_assessment_recommended': True},
    location=None,
))
case_3.status, case_3.safety_message, missing_location_provider.calls
```
