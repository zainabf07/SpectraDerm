# Module Z: Evidence Agent
Controlled synthetic data only. Model output is not medical evidence; retrieved evidence is passed to Module W for a grounded explanation.

```python
from spectraderm.agents.evidence_agent import EvidenceAgent, EvidenceAgentInput
from spectraderm.rag.llm import DeterministicMockLLM
from spectraderm.rag.rag_pipeline import RAGPipeline
from spectraderm.rag.retriever import RetrievalResult

class SyntheticRetriever:
    def retrieve(self, query, top_k=3):
        return [RetrievalResult('synthetic:chunk-0001', 'synthetic', 'Synthetic retrieved evidence text.', 1.0, 'Synthetic source', 'Synthetic organization', 'https://example.test', 'synthetic context')][:top_k]

module_w = RAGPipeline(SyntheticRetriever(), DeterministicMockLLM('Mock explanation based only on supplied retrieved evidence.'))
agent = EvidenceAgent(module_w)
vision_finding = 'Model-derived finding: pigmentation-related feature change detected.'
monitoring = {'trend': 'increasing_change', 'current_status': 'changed'}
```

```python
result = agent.interpret(EvidenceAgentInput(structured_finding=vision_finding, monitoring_result=monitoring, top_k=3))
print('MODEL-DERIVED FINDING:', result.finding_summary)
print('MONITORING CONTEXT:', result.monitoring_context)
print('EVIDENCE QUERY:', result.query)
print('RETRIEVED EVIDENCE:', result.retrieved_evidence)
print('SOURCE/CHUNK IDs:', [(item.source.source_id, item.chunk_id) for item in result.retrieved_evidence])
print('GROUNDING STATUS:', result.grounding_status)
print('GENERATED EXPLANATION:', result.explanation)
print('SAFETY/LIMITATIONS:', result.safety_message, result.limitations)
print('\nMODEL OUTPUT != MEDICAL EVIDENCE; RETRIEVED EVIDENCE -> GROUNDED EXPLANATION')
```
