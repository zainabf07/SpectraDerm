# Module AB: Product Agent
Synthetic examples only. General OTC skincare category suggestions based on a model-derived appearance pattern — not a diagnosis or treatment plan.

```python
from spectraderm.agents.product_agent import ProductAgent, ProductAgentInput

agent = ProductAgent()
cases = {
    'Pigmentation': ProductAgentInput(detected_pattern='pigmentation', safety_result={'professional_assessment_recommended': False}, monitoring_result={'trend': 'stable'}),
    'Redness/irritation': ProductAgentInput(detected_pattern='redness_irritation', safety_result={'professional_assessment_recommended': False}, monitoring_result={'trend': 'stable'}),
    'Acne-like/texture': ProductAgentInput(detected_pattern='acne_like_texture', safety_result={'professional_assessment_recommended': False}, monitoring_result={'trend': 'changed'}),
    'AA referral gate': ProductAgentInput(detected_pattern='pigmentation', safety_result={'professional_assessment_recommended': True}, monitoring_result={'trend': 'increasing_change'}),
}
```

```python
for name, agent_input in cases.items():
    result = agent.recommend(agent_input)
    print('\nCASE:', name)
    print('Model-derived pattern:', agent_input.detected_pattern)
    print('Monitoring context:', result.monitoring_context)
    print('AA safety gate:', result.safety_gate)
    print('Detected pattern:', result.detected_pattern)
    print('Recommendations:', result.recommendations)
    print('Reasons:', result.recommendation_reasons)
    print('Limitations:', result.limitations)
    print('General OTC skincare category — not a diagnosis or treatment plan.')
```
